```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense, Dropout
```

```python
df = pd.read_csv(r"C:\Users\Amanna Negi\Downloads\temp.csv")
```

```python
df.head()
```

```python
data = df[['Temperature']].values
```

```python
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(data)
```

```python
X=[]
y=[]
```

```python
window=3

for i in range(window,len(scaled_data)):
    X.append(scaled_data[i-window:i])
    y.append(scaled_data[i])
```

```python
X=np.array(X)
y=np.array(y)
```

```python
model = Sequential([
    GRU(50, return_sequences=True, input_shape=(X.shape[1],1)),
    Dropout(0.2),
    GRU(50),
    Dropout(0.2),
    Dense(1)
])
```

```python
model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)
```

```python
model.fit(
    X,
    y,
    epochs=50,
    batch_size=8
)
```

```python
def predict_temperature(values):

    data=np.array(values).reshape(-1,1)

    scaled=scaler.transform(data)

    X_test=np.array([scaled[-3:]])

    prediction=model.predict(X_test)

    result=scaler.inverse_transform(prediction)

    return result[0][0]
```

```python
predict_temperature([26,27,28])
```

```python
import pickle

with open("gru_model.pkl","wb") as file:
    pickle.dump(model,file)


with open("scaler.pkl","wb") as file:
    pickle.dump(scaler,file)
```
