```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import MinMaxScaler

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
```

```python
df = pd.read_csv(r"C:\Users\Amanna Negi\Downloads\data.csv")
```

```python
df.head()
```

```python
data = df[['Close']].values
```

```python
scaler = MinMaxScaler()
data_scaled = scaler.fit_transform(data)
```

```python
X = []
y = []
```

```python
window = 3

for i in range(window, len(data_scaled)):
    X.append(data_scaled[i-window:i])
    y.append(data_scaled[i])
```

```python
X = np.array(X)
y = np.array(y)
```

```python
model = Sequential([
    LSTM(50, return_sequences=True, input_shape=(X.shape[1], 1)),
    Dropout(0.2),
    LSTM(50),
    Dropout(0.2),
    Dense(1)
])
```

```python
model.compile(
    optimizer="adam",
    loss="mean_squared_error"
)
```

```python
model.fit(
    X,
    y,
    epochs=50,
    batch_size=8
)
```

```python
def predict_price(last_prices):

    data = np.array(last_prices).reshape(-1,1)

    scaled_data = scaler.transform(data)

    X_test = np.array([scaled_data[-3:]])

    prediction = model.predict(X_test)

    price = scaler.inverse_transform(prediction)

    return price[0][0]
```

```python
predict_price([119,121,124])
```

```python
import pickle


with open("lstm_model.pkl", "wb") as file:
    pickle.dump(model, file)


with open("scaler.pkl", "wb") as file:
    pickle.dump(scaler, file)
```
