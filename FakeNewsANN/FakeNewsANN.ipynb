```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

from sklearn.metrics import accuracy_score, classification_report
```

```python
df=pd.read_csv(r"C:\Users\Amanna Negi\Downloads\News.csv")
```

```python
df.head()
```

```python
df['label'].value_counts()
```

```python
X = df['text']
y = df['label']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.22,random_state=42,stratify=y)
```

```python
tfidf = TfidfVectorizer(max_features=5000)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)
```

```python
X_train_tfidf = X_train_tfidf.toarray()
X_test_tfidf = X_test_tfidf.toarray()
```

```python
X_train_tfidf.shape
```

```python
model = Sequential([
    Dense(128, activation='relu',
          input_shape=(X_train_tfidf.shape[1],)),
    
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    X_train_tfidf,
    y_train,
    epochs=20,
    batch_size=8,
    validation_split=0.2
)
```

```python
news = ["Scientists discover a magical pill that can make humans live forever"]


news_vector = tfidf.transform(news).toarray()


prediction = model.predict(news_vector)


if prediction[0][0] > 0.5:
    print("Fake News")
else:
    print("Real News")
```

```python
import pickle


model.save("fake_news_ann_model.keras")


with open("fake_news_tfidf.pkl","wb") as f:
    pickle.dump(tfidf,f)


print("Model Saved")
```
